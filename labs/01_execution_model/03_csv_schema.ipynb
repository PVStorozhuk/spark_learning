# Experiment 3 — reading a CSV with and without a schema

Spark UI: http://localhost:4040 → **Jobs**. Refresh after every cell. No action is called anywhere in this notebook until the last section.

```python
from pyspark.sql import SparkSession, functions as F

spark = SparkSession.builder.appName("m1-exp3-csv").getOrCreate()
path = "/opt/data/m1/orders_csv"
```

## Read A — header only, no schema, `inferSchema` left at its default

**PREDICT:** how many jobs after this cell?

```python
a = spark.read.option("header", True).csv(path)
```

## Read B — `inferSchema=True`

**PREDICT:** how many *new* jobs after this cell? How many tasks in each?

```python
b = spark.read.option("header", True).option("inferSchema", True).csv(path)
```

## Read C — explicit schema

**PREDICT:** how many *new* jobs after this cell?

```python
schema = "order_id BIGINT, customer_id INT, shop_id INT, amount DOUBLE, order_date DATE"
c = spark.read.option("header", True).schema(schema).csv(path)
```

## Compare the three schemas

**PREDICT:** is `printSchema()` an action? How do the three schemas differ?

```python
a.printSchema()
b.printSchema()
c.printSchema()
```
